```python
import skmob2
import pandas as pd
import polars as pl
from skmob2.core import TrajectoryDataFrame
import time
```

# Evaluate fast path for single biggest user

```python
path = "../../../tests/shared/data/loc-brightkite_totalCheckins.txt.gz"
df = pd.read_csv(path, sep='\t', header=None, names=['datetime', 'lat', 'lon', 'uidabc'])
tdf = TrajectoryDataFrame(df)
tdf
```

```python
times = []

for i in range(5):
    time_start = time.perf_counter()
    tdf.jump_lengths()
    time_end = time.perf_counter()
    times.append(time_end - time_start)

print(f"Mean Time taken: {sum(times) / len(times):.5f} seconds")
```
