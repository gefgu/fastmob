```python
import time

import fastmob
import pandas as pd
import polars as pl
from fastmob.core import TrajDataFrame
from fastmob.data import load_dataset
```

# Evaluate fast path for single user polars

```python
path = "../../../tests/shared/data/loc-brightkite_totalCheckins.txt.gz"
df = pl.read_csv(
    path,
    separator="\t",
    has_header=True,
    new_columns=["uid", "datetime", "lat", "lng", "location id"],
)
tdf = TrajDataFrame(df, sort=True)
tdf
```

```python
times = []

for i in range(5):
    time_start = time.perf_counter()
    tdf.jump_lengths()
    time_end = time.perf_counter()
    times.append(time_end - time_start)

print(f"Mean Time taken: {sum(times) / len(times):.5f} seconds")
```
